## STEP 5: Convert to Python

### Foundation Data Processing Functions

````python
# ============================================================
# MAITA FINANCE SMART BUDDY
# Purpose: Analyse spending, give budget advice, and calculate savings goals
# ============================================================

import re # Import the regular expression module

def clean_amount(value):
    """Converts money values like 'Rs 1,200', '$300', or '-50' into numbers."""
    if pd.isna(value):
        return None

    value = str(value)
    # Modified regex to remove commas along with other non-numeric characters (except for dot and hyphen)
    value = re.sub(r"[^\d.-]", "", value)

    try:
        return float(value)
    except ValueError:
        return None

def load_transactions(file):
    """Loads and prepares uploaded transaction CSV."""
    if file is None:
        return None, "Please upload your transaction CSV first."

    try:
        df = pd.read_csv(file.name)
    except Exception:
        return None, "I could not read this file. Please upload a valid CSV file."

    df.columns = df.columns.str.strip()

    required_columns = ["Category", "Amount"]
    missing = [col for col in required_columns if col not in df.columns]

    if missing:
        return None, f"Your CSV is missing: {missing}. It must include Category and Amount."

    df["Amount_Clean"] = df["Amount"].apply(clean_amount)
    df = df.dropna(subset=["Amount_Clean"])

    if df.empty:
        return None, "No valid amounts were found in your CSV."

    return df, "Your transactions were uploaded successfully."
````

````python
def analyse_spending(df):
    """Creates spending summary and category analysis."""

    # Separate spending and refunds
    spending = df[df["Amount_Clean"] > 0]
    refunds = df[df["Amount_Clean"] < 0]

    # Calculate totals
    total_spending = spending["Amount_Clean"].sum()
    total_refunds = refunds["Amount_Clean"].sum()
    transaction_count = len(df)

    # Group spending by category
    category_table = (
        spending.groupby("Category")["Amount_Clean"]
        .agg(["sum", "mean", "count"])
        .reset_index()
        .rename(columns={
            "sum": "Total Spent",
            "mean": "Average Spend",
            "count": "Transactions"
        })
        .sort_values("Total Spent", ascending=False)
    )

    # Highest spending category
    if category_table.empty:
        top_category = "None"
        top_amount = 0
    else:
        top_category = category_table.iloc[0]["Category"]
        top_amount = category_table.iloc[0]["Total Spent"]

    # Dashboard summary
    summary = f"""
# Maita Finance Smart Buddy Dashboard

### Total Spending: Rs {total_spending:,.2f}
### Refunds / Money Back: Rs {abs(total_refunds):,.2f}
### Number of Transactions: {transaction_count}
### Biggest Spending Area: {top_category} — Rs {top_amount:,.2f}

---

## Smart Insight
Your biggest spending category is **{top_category}**.

This category is where most of your money is going.

## Recommendation
Try reducing spending in **{top_category}** by 10–15% next month.
"""

    return summary, category_table
````

Once your foundation data processing is solid, integrate advanced AI features using hands-on-ai.

## Chat Interface Integration

````python
def money_chatbot(question, df):
    """Simple finance chatbot."""

    if not question or question.strip() == "":
        return "Ask me something about your spending or savings."

    q = question.lower()

    if df is not None:
        _, category_table = analyse_spending(df)

        if not category_table.empty:
            top_category = category_table.iloc[0]["Category"]
            top_amount = category_table.iloc[0]["Total Spent"]
            total_spending = category_table["Total Spent"].sum()
        else:
            top_category = "None"
            top_amount = 0
            total_spending = 0

    if "highest" in q or "biggest" in q:
        return f"""
Your highest spending category is **{top_category}**.

You spent **Rs {top_amount:,.2f}** there.
"""

    if "save" in q or "saving" in q:
        return """
Savings Advice:

- Reduce unnecessary spending
- Set a monthly savings target
- Track your expenses weekly
"""

    if "budget" in q:
        return """
Budget Advice:

- Separate needs and wants
- Review spending weekly
- Keep emergency savings
"""

    if "total" in q:
        return f"""
Your total spending is:

# Rs {total_spending:,.2f}
"""

    return """
I can help with:
- spending analysis
- savings advice
- budget advice
- highest spending category
"""
````

## RAG System for Financial Documents

````python
# Advanced Feature: RAG-style CSV Retrieval

def retrieve_finance_context(question, df):
    """
    Retrieves relevant information from the uploaded CSV transaction data.
    This supports the chatbot by using the user's own CSV as context.
    """

    if df is None:
        return "No transaction data has been uploaded yet."

    question = question.lower()

    spending = df[df["Amount_Clean"] > 0]
    refunds = df[df["Amount_Clean"] < 0]

    category_totals = spending.groupby("Category")["Amount_Clean"].sum().sort_values(ascending=False)

    if category_totals.empty:
        return "No spending categories were found in the uploaded data."

    if "highest" in question or "biggest" in question:
        top_category = category_totals.index[0]
        top_amount = category_totals.iloc[0]
        return f"The retrieved CSV context shows that the highest spending category is {top_category}, with Rs {top_amount:,.2f} spent."

    if "refund" in question:
        total_refunds = refunds["Amount_Clean"].sum()
        return f"The retrieved CSV context shows total refunds of Rs {abs(total_refunds):,.2f}."

    if "total" in question or "spending" in question:
        total_spending = spending["Amount_Clean"].sum()
        return f"The retrieved CSV context shows total positive spending of Rs {total_spending:,.2f}."

    return "The uploaded CSV contains transaction dates, categories, amounts, and descriptions that can be retrieved for finance analysis."
````

## Custom Financial Tools

````python
import math

def savings_calculator(goal_amount, monthly_saving):
    """Calculates savings timeline."""

    try:
        goal_amount = float(goal_amount)
        monthly_saving = float(monthly_saving)

        if goal_amount <= 0:
            return "Goal amount must be greater than 0."

        if monthly_saving <= 0:
            return "Monthly saving amount must be greater than 0."

        months = math.ceil(goal_amount / monthly_saving)

        return f"""
To save Rs {goal_amount:,.2f}
while saving Rs {monthly_saving:,.2f} per month:

You need approximately {months} months.
"""

    except:
        return "Please enter valid numbers."
        stored_df = None
````

## Gradio UI Integration

````python
import matplotlib.pyplot as plt
import gradio as gr

def create_spending_chart(category_table):
    """Creates a pie chart of spending by category."""
    if category_table.empty:
        fig, ax = plt.subplots()
        ax.text(0.5, 0.5, "No data to display", horizontalalignment='center', verticalalignment='center', transform=ax.transAxes)
        ax.axis('off')
        return fig

    fig, ax = plt.subplots(figsize=(8, 8))
    ax.pie(
        category_table["Total Spent"],
        labels=category_table["Category"],
        autopct="%1.1f%%",
        startangle=90
    )
    ax.axis("equal")  # Equal aspect ratio ensures that pie is drawn as a circle.
    ax.set_title("Spending Distribution by Category")
    return fig


def analyse_uploaded_file(file):
    """
    Handles the file upload, loads transactions, analyses spending,
    and returns the summary, table, chart, and the processed dataframe.
    """
    df, message = load_transactions(file)
    if df is None:
        return message, pd.DataFrame(), None, None # Return empty dataframe and no chart on error, and None for df

    summary, category_table = analyse_spending(df)
    chart = create_spending_chart(category_table)
    return summary, category_table, chart, df # Return the dataframe for storage in state


def ask_assistant(question, df):
    """
    Calls the money_chatbot with the current question and the stored dataframe.
    """
    return money_chatbot(question, df)


with gr.Blocks(
    title="Maita Finance Smart Buddy",
    theme=gr.themes.Soft()
) as app:

    stored_df = gr.State(None) # To store the dataframe after upload

    gr.Markdown("""
# Maita Finance Smart Buddy

Smart money decisions start with clear insights.

Upload your transactions, understand your spending habits, and plan your savings goals with confidence.
""")

    with gr.Tab("Spending Dashboard"):
        gr.Markdown("## Upload your transaction file")

        file_input = gr.File(label="Upload CSV file")
        analyse_button = gr.Button("Analyse My Spending")

        summary_output = gr.Markdown()
        table_output = gr.Dataframe(label="Spending Breakdown")
        chart_output = gr.Plot(label="Spending Chart")

        analyse_button.click(
            analyse_uploaded_file,
            inputs=file_input,
            outputs=[summary_output, table_output, chart_output, stored_df]
        )

    with gr.Tab("Ask Smart Buddy"):
        gr.Markdown("## Ask a finance question")

        question_input = gr.Textbox(
            label="Your question",
            placeholder="Example: What is my biggest spending category?"
        )

        ask_button = gr.Button("Ask")
        answer_output = gr.Markdown()

        ask_button.click(
            ask_assistant,
            inputs=[question_input, stored_df],
            outputs=answer_output
        )

    with gr.Tab("Savings Planner"):
        gr.Markdown("## Plan your savings goal")

        goal_input = gr.Number(label="How much do you want to save?")
        monthly_input = gr.Number(label="How much can you save per month?")
        calculate_button = gr.Button("Calculate My Timeline")

        savings_output = gr.Markdown()

        calculate_button.click(
            savings_calculator,
            inputs=[goal_input, monthly_input],
            outputs=savings_output
        )

    with gr.Tab("About"):
        gr.Markdown("""
## About Maita Finance Smart Buddy

**Maita Finance Smart Buddy** is a Smart Finance Assistant that helps users:

- Upload transaction data
- Clean and analyse spending
- View spending by category
- Identify the highest spending area
- Ask finance-related questions
- Calculate savings timelines

## CSV Requirements

Your CSV must include:

- `Category`
- `Amount`

Optional columns:

- `Date`
- `Description`

## Example Questions

- What is my biggest spending category?
- How can I save more?
- Give me budget advice.
- What is my total spending?
""")

app.launch(share=True)
````
