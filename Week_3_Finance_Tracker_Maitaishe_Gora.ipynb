````python
amount = int(input("Enter expense amount: "))
if amount >= 100:
    category = "Major Expense"
    print("Reduce your expenses, it's too much")
elif amount >= 50:
    category = "Moderate Expense"
    print("Manage your expenses, you are at borderline")
elif amount >= 10:
    category = "Minor Expense"
    print("You are doing great, save more")
else:
    category = "Small Purchase"
    print("You are the queen of saving money")
````
