```python
print("=== Personal Finance Tracker v0.1 ===")
print("Week 3: Smart Budget Decisions\n")

# User setup
name = input("Enter your name: ")
monthly_income = float(input("Enter your monthly income: $"))

print(f"\nWelcome to your finance tracker, {name}!")
print(f"Monthly income: ${monthly_income:.2f}\n")

# This week's feature: Expense categorisation
expense_amount = float(input("Enter an expense: $"))
expense_description = input("What was this expense for? ").lower()

# Simple categorisation
if any(word in expense_description for word in ["rent", "mortgage", "house", "apartment"]):
    category = "Housing"
elif any(word in expense_description for word in ["food", "groceries", "supermarket", "restaurant", "coffee"]):
    category = "Food"
elif any(word in expense_description for word in ["transport", "bus", "train", "uber", "fuel"]):
    category = "Transportation"
elif any(word in expense_description for word in ["entertainment", "movies", "netflix", "game", "concert"]):
    category = "Entertainment"
elif any(word in expense_description for word in ["clothes", "shopping", "fashion", "shoes"]):
    category = "Shopping"
else:
    category = "Other"

# Budget percentage calculation
expense_percent = (expense_amount / monthly_income) * 100

# Feedback logic
if expense_percent > 30:
    feedback = "Whoa! That’s a big chunk of your income. Maybe rethink this expense."
elif expense_percent > 10:
    feedback = "Moderate spend. Keep an eye on your budget."
else:
    feedback = "Small expense, easy to manage."

# Output results
print(f"\nExpense Summary:")
print(f"- Description: {expense_description.capitalize()}")
print(f"- Category: {category}")
print(f"- Amount: ${expense_amount:.2f} ({expense_percent:.1f}% of your monthly income)")
print(f"- Feedback: {feedback}\n")

print("Keep tracking! Your future self will thank you. 🚀")
```
